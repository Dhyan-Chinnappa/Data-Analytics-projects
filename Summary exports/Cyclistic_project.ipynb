```python
#Step 1: Load and trim ghost columns(my file has 16,300 empty extra columns)
df = pd.read_csv('C:/Users/dhyan/OneDrive/Desktop/Data Analytics Project folder/Cyclistic trip data/Summary exports/cyclistic_combined.csv',low_memory = False)
df = df[df.columns[:15]]  #keeps only real 15 columns
```

```python
#Step 2: Convert ride_length(HH:MM:SS text) to seconds
df['ride_length_sec'] = pd.to_timedelta(df['ride_length']).dt.total_seconds()
```

```python
#Step 3: Group by member_casual and calculate all metrics at once 
summary = df.groupby('member_casual').agg(
    mean_ride_length=('ride_length_sec','mean'),
    max_ride_length=('ride_length_sec','max'),
    median_ride_length=('ride_length_sec','median'),
    mode_day_of_week=('day_of_week', lambda x: x.mode()[0]),
    total_rides = ('ride_length_sec','count')
)
```

```python
#Step 4: Convert seconds back to HH:MM:SS for redable output
for col in ['mean_ride_length','max_ride_length','median_ride_length']:
    summary[col] = pd.to_timedelta(summary[col],unit='s')
print(summary)
```
